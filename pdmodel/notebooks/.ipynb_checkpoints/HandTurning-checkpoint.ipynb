```python
path1 = r'../../handOutput/*_B*_hand.txt'
files_ls1 = glob.glob(path1)

path2 = r'../../handOutput2/*_B*_hand.txt'
files_ls2 = glob.glob(path2)

path3 = r'../../handOutput3/*_B*_hand.txt'
files_ls3 = glob.glob(path3)
files_ls3
```

```python
output_dt = {}
for files in files_ls3:
    
    dt = joblib.load(files)

    output_dt[files] = cal_hand_turn_features(dt)
```

```python
df = pd.DataFrame.from_dict(output_dt).T
df = df.reset_index()
df.columns = ["file_path","Thumb mean clarity", "Thumb 0.2 clarity", "Thumb 0.4 clarity", "Thumb 0.6 clarity",
              "Pinky mean clarity", "Pinky 0.2 clarity", "Pinky 0.4 clarity", "Pinky 0.6 clarity",
              "Hand turning duration"]
df["Date"] = df.file_path.apply(lambda x: x.split("/")[-1].split("_")[0])
df["Date"] = df["Date"].apply(lambda x: x[0:4] + "-" + x[4:6] + "-" + x[6:])
df["PID"] = df.file_path.apply(lambda x: int(x.split("/")[-1].split("_")[1]))
df["Hand_id"] = df.file_path.apply(lambda x: x.split("/")[-1].split("_")[2][-1])
df = df.set_index(["Date", "PID"])


```

```python
h_g = df.groupby("Hand_id")

col = ["Thumb mean clarity", "Thumb 0.2 clarity", "Thumb 0.4 clarity", "Thumb 0.6 clarity",
        "Pinky mean clarity", "Pinky 0.2 clarity", "Pinky 0.4 clarity", "Pinky 0.6 clarity",
        "Hand turning duration"]

left_df = h_g.get_group("L").loc[:, col]

left_df.columns = ["Left thumb mean clarity", "Left thumb 0.2 clarity", "Left thumb 0.4 clarity", "Left thumb 0.6 clarity",
        "Left pinky mean clarity", "Left pinky 0.2 clarity", "Left pinky 0.4 clarity", "Left pinky 0.6 clarity",
        "Left hand turning duration"]

right_df = h_g.get_group("R").loc[:, col]

right_df.columns = ["Right thumb mean clarity", "Right thumb 0.2 clarity", "Right thumb 0.4 clarity", "Right thumb 0.6 clarity",
        "Right pinky mean clarity", "Right pinky 0.2 clarity", "Right pinky 0.4 clarity", "Right pinky 0.6 clarity",
        "Right hand turning duration"]
```

```python


all_df = pd.read_csv("/HDD3/Leo/Pose2D/original_pd.csv").iloc[:-2, :]
all_df["Date"] = all_df["Date"].apply(lambda x: str(int(x))[:4]+ "-" + str(int(x))[4:6] + "-" + str(int(x))[-2:])
all_df["PID"] = all_df["PID"].astype(int)
all_df = all_df.set_index(["Date", "PID"])
all_df["medcine"] = all_df["medcine"].fillna(0)

info_df = all_df.loc[:, ["AGE", "Gender", "PD Stages"]]
info_df.columns = ["Age", "Gender", "PD Stages"]
```

```python
handturning_df = pd.concat([left_df, right_df, info_df], axis=1)
handturning_df = handturning_df.dropna()
handturning_df["PD Stages"] = handturning_df["PD Stages"].apply(lambda x: "Normal" if x == 0 else "Early-PD" if x in [1,2] else "Advanced-PD")
```

```python
norm_hand = handturning_df[handturning_df["PD Stages"] == "Advanced-PD"]
norm_hand[norm_hand['Left thumb mean clarity'] < 0.2]
```

```python
ls = ['Left thumb mean clarity', 
       'Left pinky mean clarity', 
       'Left hand turning duration', 'Right thumb mean clarity',
    'Right pinky mean clarity','Right hand turning duration']

combinations_2d = list(combinations(ls, 2))
df = handturning_df

for i, combo in enumerate(combinations_2d):
    fig = px.scatter(df, x=combo[0], y=combo[1], color='PD Stages', trendline='ols',
                     trendline_scope='trace', color_discrete_sequence=['#10439F', "#E19898", "#AC2A2A"])
    fig.update_traces(marker=dict(size=15))
    fig.update_xaxes(title_text=combo[0], gridcolor='black', showgrid=True, visible=True, linecolor='black', linewidth=2)  # Make x-axis black and thicker
    fig.update_yaxes(title_text=combo[1], gridcolor='black', showgrid=True, visible=True, linecolor='black', linewidth=2)  # Make y-axis black and thicker

    fig.update_layout(showlegend=False, legend_tracegroupgap=300, legend={'traceorder': 'reversed'}, plot_bgcolor='#FFFFFF', 
                      font=dict(family="Serif", size=36))
    fig.update_layout(height=800, width=800)
    fig.show()
```
