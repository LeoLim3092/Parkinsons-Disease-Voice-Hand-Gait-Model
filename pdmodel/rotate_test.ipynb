```python
import cv2
import mediapipe as mp
import time
import numpy as np
from matplotlib import pyplot as plt
# 初始化 Mediapipe Hands 模組和繪圖工具
mp_hands = mp.solutions.hands
mp_drawing = mp.solutions.drawing_utils

# 打開影片文件（替換為您的影片路徑）
video_path = '/HDD3/PD_Data/TW/PD_Data_Hand/20200716/20200716_19BL.mp4'  # 請替換為你自己的影片路徑
cap = cv2.VideoCapture(video_path)

# 創建 VideoWriter 用於保存辨識後的影片
output_path = 'output_video.mp4'
frame_width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
frame_height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
fps = int(cap.get(cv2.CAP_PROP_FPS))
fourcc = cv2.VideoWriter_fourcc(*'mp4v')
out = cv2.VideoWriter(output_path, fourcc, fps, (frame_width, frame_height))

# 計算程式執行時間
start_time = time.time()
out_dt = {}

# 初始化 Mediapipe Hands 模組
with mp_hands.Hands(static_image_mode=False, max_num_hands=2, min_detection_confidence=0.5) as hands:
    i=0
    while cap.isOpened():
        ret, frame = cap.read()
        
        if not ret:
            break

        # 將 BGR 顏色空間轉換為 RGB
        frame_rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)

        # 使用 Mediapipe Hands 進行手勢檢測
        results = hands.process(frame_rgb)
        
        all_hand = []
        
        # 如果檢測到手部
        if results.multi_hand_landmarks:
            for h_idx, hand_landmarks in enumerate(results.multi_hand_landmarks):
                hand_ls = []
                for l_idx, landmark in enumerate(hand_landmarks.landmark):
                    hand_ls.append([landmark.x, landmark.y, landmark.z])

                hand_arr = np.array(hand_ls)
                all_hand.append(hand_arr)
                break
            i += 1
            out_dt[f"{i}"] = all_hand
                
        out.write(frame)

        # 顯示影像
        # cv2.imshow('Hand Gesture Recognition', frame)

        # # 將處理後的影像寫入輸出影片
        

        # # 按 'q' 鍵退出
        # if cv2.waitKey(10) & 0xFF == ord('q'):
        #     break


four_landmark_x = []

for frame in out_dt:
    four_landmark_x.append(out_dt[frame][0][4][0])

print(four_landmark_x)
# 計算並顯示總執行時間
end_time = time.time()
execution_time = end_time - start_time
print(f"Total execution time: {execution_time:.2f} seconds")

# 釋放資源
cap.release()
out.release()
cv2.destroyAllWindows()

plt.figure(figsize=(10, 6))
plt.plot(range(len(four_landmark_x)), four_landmark_x, label='four_landmark_x')
        
plt.xlabel('Frame Index')
plt.ylabel('four_landmark_x')
plt.title('four_landmark_x')
plt.legend()
plt.grid(True)
plt.show()
```

```python
non_zero_var = four_landmark_x

t = np.arange(len(non_zero_var))

# 歸一化 x 軸數據到 0 到 1
t_normalized = (t - t.min()) / (t.max() - t.min())

# 進行多項式擬合
degree = 10  
p = np.polyfit(t_normalized, non_zero_var, degree)

print(p)
# 使用擬合結果計算趨勢
trend = np.polyval(p, t_normalized)

plt.figure(figsize=(10, 6))
plt.plot(range(len(non_zero_var)), non_zero_var , label='Point 4 x coordinate')
plt.plot(range(len(trend)), trend, label='trend')
        
plt.xlabel('Frame Index')
plt.ylabel('Point 4 x coordinate')
plt.title('Point 4 x coordinate Over Time')
plt.legend()
plt.grid(True)
plt.show()
```

```python
detrend = non_zero_var - trend

plt.figure(figsize=(10, 6))
plt.plot(range(len(detrend)), detrend, label='Point 4 x coordinate')
        
plt.xlabel('Frame Index')
plt.ylabel('Point 4 x coordinate')
plt.title('Point 4 x coordinate Over Time')
plt.legend()
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# 假設你已經有座標數據 `x`
# Step 1: 計算速度 (一階差分)
x = detrend  # 替換為你的數據
velocity = np.diff(x)

# Step 2: 傅里葉轉換
fft_result = np.fft.fft(velocity)
frequencies = np.fft.fftfreq(len(velocity))

# Step 3: 計算頻譜能量
power_spectrum = np.abs(fft_result)**2

# Step 4: 提取高頻分量比例
high_freq_threshold = 0.1  # 設置高頻分量的閾值
high_freq_indices = np.abs(frequencies) > high_freq_threshold
high_freq_energy = np.sum(power_spectrum[high_freq_indices])
total_energy = np.sum(power_spectrum)
high_freq_ratio = high_freq_energy / total_energy

# === 計算頻譜能量的變異數 ===
power_variance = np.var(power_spectrum)
print(f"頻譜能量的變異數: {power_variance:.2f}")

# 結果輸出
print(f"高頻分量能量比例: {high_freq_ratio:.2f}")

# 正規化功率譜
power_spectrum_norm = power_spectrum / np.sum(power_spectrum)

# 頻譜熵
spectral_entropy = -np.sum(power_spectrum_norm * np.log(power_spectrum_norm))

# 頻譜重心
spectral_centroid = np.sum(frequencies * power_spectrum) / np.sum(power_spectrum)

# 頻譜變異數
spectral_variance = np.sum(((frequencies - spectral_centroid)**2) * power_spectrum) / np.sum(power_spectrum)

# 高頻能量比例
threshold = 0.1  # 假設紅線頻率閾值為 0.1
high_frequency_energy_ratio = np.sum(power_spectrum[frequencies > threshold]) / np.sum(power_spectrum)

print(f"Spectral Entropy: {spectral_entropy}")
print(f"Spectral Centroid: {spectral_centroid}")
print(f"Spectral Variance: {spectral_variance}")
print(f"High-Frequency Energy Ratio: {high_frequency_energy_ratio}")

# Step 5: 計算最大功率對應的頻率
max_power_index = np.argmax(power_spectrum)  # 找到最大功率對應的索引
max_power_frequency = frequencies[max_power_index]  # 獲取對應的頻率

print(f"The frequency with maximum power is: {max_power_frequency}")

# Step 5: 視覺化
plt.figure(figsize=(12, 6))
plt.plot(np.abs(frequencies), power_spectrum, label="Power Spectrum")
plt.axvline(high_freq_threshold, color='r', linestyle='--', label="High Frequency Threshold")
plt.xlabel("Frequency")
plt.ylabel("Power")
plt.title("Frequency Spectrum of Velocity")
plt.legend()
plt.show()
```
